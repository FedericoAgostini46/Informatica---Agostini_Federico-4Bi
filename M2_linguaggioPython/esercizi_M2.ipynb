## ESERCIZIO 1 - Durata di una sessione di laboratorio
Scrivere nella cella dell'esercizio 1 il programma che legge in ingresso la durata di una sessio‐
ne di laboratorio espressa in secondi e la stampa nel formato ore, minuti e secondi, usando
soltanto la divisione intera e il resto. Il programma stampa anche la stessa durata espressa in
ore decimali con due cifre dopo la virgola, e segnala con un messaggio il caso in cui la durata
superi le quattro ore.

```python
secondi = int(input("Durata in secondi: "))

ore = secondi // 3600
resto = secondi % 3600
minuti = resto // 60
sec = resto % 60

print(f"{ore:02d}:{minuti:02d}:{sec:02d}")
print(f"in ore decimali: {secondi / 3600:.2f}")

if secondi > 4 * 3600:
    print("attenzione: la sessione supera le quattro ore")
```

## ESERCIZIO 2 - Media dei voti e giudizio
Scrivere nella cella dell'esercizio 2 il programma che legge in ingresso il nome di uno studen‐
te, tre voti in virgola mobile e le ore di assenza, calcola la media dei tre voti e stampa il giudizio
secondo questa regola: se le ore di assenza superano 50 il giudizio è non valutabile , indi‐
pendentemente dalla media; altrimenti il giudizio è ottimo da 8 in su, sufficiente da 6 in
su, insufficiente sotto il 6. La media va stampata con due decimali dentro una f-string,
insieme al nome dello studente.

```python
nome = input("Nome: ")
v1 = float(input("Voto 1: "))
v2 = float(input("Voto 2: "))
v3 = float(input("Voto 3: "))
assenze = float(input("Ore di assenza: "))

media = (v1 + v2 + v3) / 3

if assenze > 50:
    giudizio = "non valutabile"
elif media >= 8:
    giudizio = "ottimo"
elif media >= 6:
    giudizio = "sufficiente"
else:
    giudizio = "insufficiente"

print(f"{nome}: media {media:.2f}, giudizio {giudizio}")
```

Scrivere nella cella dell'esercizio 3 il programma che legge in ingresso un valore iniziale, un
valore finale e un passo, tutti interi, e stampa la tavola di conversione da chilometri a miglia e a
metri per tutti i valori dell'intervallo, estremi compresi. Un chilometro corrisponde a 0,621371

## ESERCIZIO 5 - Conteggio dei caratteri di una frase
Scrivere nella cella dell'esercizio 5 il programma che legge in ingresso una frase e conta, con
un unico ciclo for sui caratteri, quante vocali, quante consonanti, quante cifre e quanti altri
caratteri contiene. Il programma stampa i quattro conteggi e la percentuale di vocali sul totale
dei caratteri, con un decimale. Le lettere accentate vanno considerate consonanti: è una
semplificazione voluta, da dichiarare in un commento.

```python
frase = input("Frase: ")

# Semplificazione voluta: le lettere accentate non sono in "aeiou",
# quindi vengono contate come consonanti.
vocali = 0
consonanti = 0
cifre = 0
altri = 0

for c in frase:
    if c.isdigit():
        cifre += 1
    elif c.isalpha():
        if c.lower() in "aeiou":
            vocali += 1
        else:
            consonanti += 1
    else:
        altri += 1

print(f"vocali: {vocali}, consonanti: {consonanti}, cifre: {cifre}, altri: {altri}")
if len(frase) > 0:
    print(f"percentuale di vocali: {vocali / len(frase) * 100:.1f}%")
else:
    print("frase vuota")
```

## Esercizio 7 - Analisi delle cifre di un numero
Scrivere nella cella dell'esercizio 7 il programma che legge in ingresso un intero non negativo e ne calcola, con un ciclo while che usa soltanto il resto e la divisione intera per dieci, la somma delle cifre, il numero di cifre e la cifra maggiore. Il numero zero va gestito correttamente: ha una cifra, la somma è zero e la cifra maggiore è zero. Non è ammesso trasformare il
numero in stringa.

```python
numero = int(input("Numero intero non negativo: "))
if numero < 0:
    print("il numero deve essere non negativo")
else:
    n = numero
    cifre = 0
    somma = 0
    maggiore = 0
    while n > 0:
        c = n % 10          # ultima cifra
        somma = somma + c
        cifre = cifre + 1
        if c > maggiore:
            maggiore = c
        n = n // 10         # elimina l'ultima cifra
    if numero == 0:         # lo zero ha una cifra: il ciclo non gira
        cifre = 1
    print("cifre:", cifre)
    print("somma delle cifre:", somma)
    print("cifra maggiore:", maggiore)
```

## Esercizio 8 - Conversione fra decimale e binario nei due sensi
Scrivere nella cella dell'esercizio 8 il programma che legge in ingresso un intero non negativo,
ne costruisce la rappresentazione binaria come stringa con il metodo dei resti successivi, e la
confronta con il risultato della funzione predefinita bin , stampando verifica superata op‐
pure verifica fallita . Il programma legge poi una seconda stringa composta di soli zeri e
uni e ne calcola il valore decimale scorrendone i caratteri da sinistra a destra, senza usare la
conversione predefinita. Se la seconda stringa contiene caratteri diversi da 0 e 1 , il pro‐
gramma lo segnala e non calcola nulla.

```python
n = int(input("Numero intero non negativo: "))
if n < 0:
    print("il numero deve essere non negativo")
else:
    # metodo dei resti successivi
    if n == 0:
        binario = "0"
    else:
        binario = ""
        resto_n = n
        while resto_n > 0:
            binario = str(resto_n % 2) + binario    # ogni resto va davanti
            resto_n = resto_n // 2
    print(n, "in binario:", binario)
    if binario == bin(n)[2:]:
        print("verifica superata")
    else:
        print("verifica fallita")

    # da binario a decimale, da sinistra a destra
    s = input("Stringa di zeri e uni: ")
    valida = s != ""
    for c in s:
        if c != "0" and c != "1":
            valida = False
    if not valida:
        print("la stringa deve contenere solo 0 e 1")
    else:
        valore = 0
        for c in s:
            valore = valore * 2 + int(c)
        print(s, "in decimale:", valore)
```

## Esercizio 9 - Ricerca sequenziale in un elenco di codici
L'elenco delle aule assegnate alle classi è memorizzato in una sola stringa, con le voci separa‐
te dal punto e virgola: è la forma con cui in Python si lavora finché non si conoscono le liste, e
corrisponde all'array di stringhe usato in Java.
Scrivere nella cella dell'esercizio 9 il programma che, a partire dalla stringa
"A101;B204;C307;D410;E512" , legge in ingresso un codice e ne cerca la presenza con una
scansione voce per voce: il programma estrae le voci una alla volta con find e slicing e le
confronta con il codice cercato, interrompendo la scansione appena lo trova. Va stampata la
posizione della voce contando da uno, oppure il messaggio di codice assente. La ricerca è in‐
sensibile agli spazi agli estremi e alle maiuscole. Il caso «non trovato» va gestito con la clauso‐
la else del ciclo, senza variabili booleane di appoggio.

```python
elenco = "A101;B204;C307;D410;E512"
codice = input("Codice da cercare: ").strip().upper()

inizio = 0
numero = 0
while inizio < len(elenco):
    fine = elenco.find(";", inizio)
    if fine == -1:                  # ultima voce: non c'è il punto e virgola
        fine = len(elenco)
    voce = elenco[inizio:fine]
    numero = numero + 1
    if voce == codice:
        print(codice, "è la voce numero", numero, "dell'elenco")
        break
    inizio = fine + 1
else:
    print(codice, "non è presente nell'elenco")
```

## Esercizio 10 - Statistiche di una sequenza letta con sentinella
Scrivere nella cella dell'esercizio 10 il programma che legge in ingresso una sequenza di voti in
virgola mobile, uno per riga, e termina quando l'utente digita 0 . Il programma calcola e stam‐
pa il numero di voti inseriti, il massimo, il minimo, la media con due decimali e quanti voti sono
risultati sufficienti, cioè maggiori o uguali a 6. I voti fuori dall'intervallo da 1 a 10 vanno
segnalati e ignorati senza interrompere la lettura. Se non viene inserito alcun voto valido il
programma lo dichiara e non calcola le statistiche.
Massimo e minimo vanno inizializzati con il primo valore valido letto, non con una costante
scelta a priori: è la stessa scelta corretta già adottata in Java.

```python
validi = 0
somma = 0
sufficienze = 0
massimo = 0
minimo = 0

voto = float(input("Voto (0 per finire): "))
while voto != 0:
    if voto < 1 or voto > 10:
        print("valore", voto, "fuori intervallo: ignorato")
    else:
        if validi == 0:             # primo voto valido: inizializza massimo e minimo
            massimo = voto
            minimo = voto
        if voto > massimo:
            massimo = voto
        if voto < minimo:
            minimo = voto
        validi = validi + 1
        somma = somma + voto
        if voto >= 6:
            sufficienze = sufficienze + 1
    voto = float(input("Voto (0 per finire): "))

if validi == 0:
    print("nessun voto valido inserito")
else:
    print("voti validi:", validi)
    print(f"massimo: {massimo:.2f}")
    print(f"minimo: {minimo:.2f}")
    print(f"media: {somma / validi:.2f}")
    print("sufficienze:", sufficienze)
```

## Esercizio 11 - Palindroma con i due indici
Scrivere nella cella dell'esercizio 11 il programma che legge in ingresso una frase, ne costrui‐
sce la versione normalizzata tenendo soltanto le lettere e portandole in minuscolo, e verifica
se è palindroma con il confronto fra due indici che si avvicinano dagli estremi verso il centro,
interrompendo il ciclo al primo carattere diverso. Lo slicing con passo negativo va usato solo
alla fine, per stampare una riga di controllo che confronta il risultato ottenuto con quello dato
dal confronto ripulita >= ripulita[>:-1] . Il programma stampa anche quanti confronti
sono stati eseguiti prima di decidere.

```python
frase = input("Frase: ")

ripulita = ""
for c in frase:
    if c.isalpha():
        ripulita = ripulita + c.lower()
print("testo normalizzato:", ripulita)

i = 0
j = len(ripulita) - 1
confronti = 0
palindroma = True
while i < j:
    confronti = confronti + 1
    if ripulita[i] != ripulita[j]:
        palindroma = False
        break
    i = i + 1
    j = j - 1

print("palindroma:", palindroma)
print("confronti eseguiti:", confronti)
print("controllo con lo slicing:", ripulita == ripulita[::-1])
```

## Esercizio 12 - Occorrenze di una sottostringa senza il metodo count
Scrivere nella cella dell'esercizio 12 il programma che legge in ingresso un testo e una sotto‐
stringa da cercare, e conta quante volte la sottostringa compare nel testo usando find ripe‐
tutamente dentro un ciclo, senza chiamare count . Il programma stampa il numero di occor‐
renze e la posizione di ciascuna, contando da zero. Alla fine stampa la stessa cosa calcolata
con count , come verifica del proprio risultato, e stampa il testo con tutte le occorrenze sosti‐
tuite da un asterisco per ciascun carattere della sottostringa, ottenuto ricostruendo la stringa
con lo slicing e la concatenazione, senza replace .
Attenzione al caso in cui la sottostringa cercata sia vuota: il programma deve rifiutarla con un
messaggio prima di iniziare la ricerca.

```python
testo = input("Testo: ")
sotto = input("Sottostringa da cercare: ")

if sotto == "":
    print("la sottostringa non può essere vuota")
else:
    occorrenze = 0
    posizioni = ""
    mascherato = ""
    inizio = 0                      # da dove riprende la copia del testo
    pos = testo.find(sotto)
    while pos != -1:
        occorrenze = occorrenze + 1
        posizioni = posizioni + str(pos) + " "
        mascherato = mascherato + testo[inizio:pos] + "*" * len(sotto)
        inizio = pos + len(sotto)
        pos = testo.find(sotto, inizio)
    mascherato = mascherato + testo[inizio:]

    print("occorrenze:", occorrenze)
    print("posizioni:", posizioni.strip())
    print("verifica con count:", testo.count(sotto))
    print("testo mascherato:", mascherato)
```
