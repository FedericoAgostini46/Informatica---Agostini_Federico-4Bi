## Esercizio guidato 1 — tipi e conversione di un dato letto come testo

La variabile `posti` contiene un numero scritto come testo, così come sarebbe arrivato da `input`.
Stampa il tipo delle tre variabili, converti `posti` in intero e stampa quanti posti risultano occupati, arrotondando all'intero con `round`.

**Risultato atteso:** `<class 'str'>`, `<class 'str'>`, `<class 'float'>` e poi `18`.

```python
codice_aula = "A12"
posti = "24"          # nel programma reale:  posti = input("Posti in aula: ")
occupazione = 0.75

# 1. Stampa il tipo delle tre variabili
print(type(codice_aula))
print(type(posti))
print(type(occupazione))

# 2. Converti 'posti' in intero, calcola i posti occupati e arrotonda
posti_int = int(posti)
posti_occupati = round(posti_int * occupazione)

# 3. Stampa il risultato finale
print(posti_occupati)
```

- **type(...):** Restituisce il tipo di dato della variabile passata come argomento (es. str per stringhe, float per numeri decimali).

- **int(posti):** Converte la stringa "24" nel numero intero 24 così da poter effettuare calcoli matematici.

- **round(...):** Funzione utile per arrotondare il risultato del calcolo (18.0) al numero intero corrispondente (18).

## Esercizio guidato 2 — dalla durata in secondi al formato orario

La variabile `secondi_totali` contiene la durata complessiva di una sessione di laboratorio.
Calcola ore, minuti e secondi usando soltanto `//` e `%`, senza convertire in `float`, e lascia che
la stampa già scritta produca il formato orario.

**Risultato atteso:** `02:34:56`.

```python
secondi_totali = 9296

# Calcolo di ore, minuti e secondi usando solo la divisione intera (//) e il modulo (%)
ore = secondi_totali // 3600
resto_secondi = secondi_totali % 3600
minuti = resto_secondi // 60
secondi = resto_secondi % 60

# Stampa formattata (già impostata per produrre ore:minuti:secondi con due cifre)
print(f"{ore:02d}:{minuti:02d}:{secondi:02d}")
```

- **secondi_totali // 3600:** Poiché un'ora è composta da 3600 secondi, dividendo per 3600 con la divisione intera otteniamo il numero di ore (9296 // 3600 = 2).

- **secondi_totali % 3600:** Il modulo ci restituisce i secondi che avanzano dopo aver estratto le ore intere (9296 % 3600 = 2096).

- **resto_secondi // 60:** Dividendo i secondi rimanenti per 60, otteniamo i minuti (2096 // 60 = 34).

- **resto_secondi % 60:** Il modulo finale ci dà i secondi residui che non completano un minuto (2096 % 60 = 56).

- **F-string con :02d:** La formattazione {:02d} garantisce che ogni valore venga stampato sempre su due cifre, aggiungendo uno zero iniziale se il numero è inferiore a 10 (es. 02, 34, 56).

## Esercizio guidato 3 — due letture di temperatura

Le due variabili contengono altrettante letture di un sensore, arrivate come testo. Convertile,
stampa la differenza con un decimale e la loro media troncata all'intero, non arrotondata.

**Risultato atteso:** `3.5` e poi `20`.

```python
prima_lettura = "18.4"     # nel programma reale arriverebbero da due input()
seconda_lettura = "21.9"

# 1. Converti le stringhe in numeri decimali (float)
t1 = float(prima_lettura)
t2 = float(seconda_lettura)

# 2. Calcola la differenza (usando abs per sicurezza o la seconda meno la prima)
differenza = abs(t2 - t1)

# 3. Calcola la media e troncala all'intero (usando la divisione intera //)
media = (t1 + t2) / 2
media_troncata = int(media)

# 4. Stampa i risultati formattati come richiesto
print(f"{differenza:.1f}")
print(media_troncata)
```

- **float(...):** Converte le stringhe contenenti i numeri con la virgola (il punto in Python) in numeri reali (float), necessari per mantenere le cifre decimali.

- **abs(t2 - t1):** Calcola il valore assoluto della differenza tra le due temperature (pari a 3.5). L'uso di f"{differenza:.1f}" garantisce la stampa esattamente con un decimale.

- **int(media):** Converte il risultato della media in un numero intero tramite int(), effettuando il troncamento (rimozione della parte decimale) anziché l'arrotondamento matematico, ottenendo così 20 da 20.15.

## Esercizio guidato 4 — una riga di registro incolonnata

Stampa una sola riga in cui il nome occupa venti colonne allineato a sinistra, la classe ne occupa sei
allineata a sinistra e la media ne occupa otto allineata a destra con due decimali.

**Risultato atteso:** `Rossi Maria          4BI       7.44` (con le colonne allineate).

```python
nome = "Rossi Maria"
classe = "4BI"
media = 7.4375

# Stampa la riga incolonnata usando una sola f-string:
# - Nome in 20 colonne allineato a sinistra ({nome:<20})
# - Classe in 6 colonne allineato a sinistra ({classe:<6})
# - Media in 8 colonne allineato a destra con due decimali ({media:>8.2f})
print(f"{nome:<20}{classe:<6}{media:>8.2f}")
```

- **{nome:<20}:** Allinea la stringa del nome a sinistra (<) riservando uno spazio complessivo di 20 caratteri.

- **{classe:<6}:** Allinea la stringa della classe a sinistra (<) riservando 6 caratteri.

- **{media:>8.2f}:** Allinea il numero a destra (>) in uno spazio di 8 caratteri, formattandolo automaticamente con due cifre decimali (.2f), il che provvede anche a effettuare il corretto arrotondamento da 7.4375 a 7.44.
